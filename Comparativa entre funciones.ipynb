# Comparación entre funciones originadas con distintas densidades de tercer orden con acoplamiento cuadrático

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetro
mu = 1.0   # puedes cambiarlo

# Dominio
r = np.linspace(0, 5, 5000)  # evitamos r=0 por potencias fraccionarias

# ----------- f3rees1 -----------
def f3rees1(r, mu):
    return (1 + r**2 - (101 * r**4) / (7**(1/3)*(-1085 * r**6 - 1134 * r**2 * mu + 18 * np.sqrt(7) * r**2
                                                 * np.sqrt(3699 * r**8+ 1085 * r**4 * mu + 567 * mu**2))**(1/3))
        + (1/7)
        * (
            -7595 * r**6
            - 7938 * r**2 * mu
            + 126 * np.sqrt(7) * r**2
            * np.sqrt(
                3699 * r**8
                + 1085 * r**4 * mu
                + 567 * mu**2
            )
        )**(1/3)
    )


# ----------- f3rees2 -----------
def f3rees2(r, mu):
    C = -2219*r**4 - 2268*mu \
        + 18*np.sqrt(14)*np.sqrt(
            15174*r**8 + 2219*r**4*mu + 1134*mu**2
        )

    return 1 + r**2 \
           - (209*r**(10/3))/(7**(1/3)*C**(1/3)) \
           + (r**(2/3)*C**(1/3))/7**(2/3)


# Evaluación
f1 = f3rees1(r, mu)
f2 = f3rees2(r, mu)
```

```python
plt.figure(figsize=(6,5))

plt.plot(r, f1, linewidth=2.0, color="#1f3b73",
         label=r'$Z_{3}$')

plt.plot(r, f2, linewidth=2.0, linestyle='--',
         color="#8b1e3f",
         label=r"$Z'_{3}$")

# Línea horizontal en y = 0
plt.axhline(0, color='gray', linestyle='--', linewidth=1.2)

plt.xlabel(r'$r$', fontsize=13)
plt.ylabel(r'$f(r)$', fontsize=13)

plt.grid(True, linestyle=':', linewidth=0.6, alpha=0.7)
plt.legend(frameon=False, fontsize=11)

plt.tick_params(direction='in')
plt.tight_layout()
plt.savefig("Comparativa entre accion1 y accion3.png", dpi=300, bbox_inches="tight")
plt.show()
```
